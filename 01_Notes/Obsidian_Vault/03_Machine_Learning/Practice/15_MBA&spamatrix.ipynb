```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules








```

```python
# 1. Load dataset
df = pd.read_csv("dataset_name.csv", header=None)
```

```python
# 2. Convert DataFrame to list of transactions
transactions = df.apply(lambda x: x.dropna().tolist(), axis=1).tolist()
```

```python
# 3. Encode transactions (creates sparse matrix internally)
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions, sparse=True)
```

```python
# 4. Create DataFrame from sparse matrix
encoded_df = pd.DataFrame.sparse.from_spmatrix(
    te_array,
    columns=te.columns_
)
```

```python

# 5. Apply Apriori algorithm
frequent_itemsets = apriori(
    encoded_df,
    min_support=0.3,
    use_colnames=True
)
```

```python
# 6. Generate association rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)
```

```python
# 7. Display results
print("Frequent Itemsets:")
print(frequent_itemsets)

print("\nAssociation Rules:")
print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```
