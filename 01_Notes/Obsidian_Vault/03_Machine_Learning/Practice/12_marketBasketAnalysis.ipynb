```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
transactions = []

with open("Datasets/groceries.csv","r") as f:
    for line in f:
        items = line.strip().split(",")
        transactions.append(items)
```

```python
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions,sparse=True)
df = pd.DataFrame.sparse.from_spmatrix(te_array,columns=te.columns_)
df
```

```python
# Apriori ->

freq_items = apriori(df, min_support = 0.02, use_colnames = True)
freq_items
```

```python
# Association rules

rules = association_rules(freq_items, metric = 'confidence', min_threshold = 0.2)
rules
```

```python
rules = rules[
    (rules['lift'] > 1) &
    (rules['confidence'] >= 0.6)
]

# Sort rules by lift
rules = rules.sort_values(by='lift', ascending=False)

print("\nAssociation Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
