```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df = sns.load_dataset("iris")
df
```

```python
X = df.drop("species",axis=1)
```

```python
Ss = StandardScaler()
X = Ss.fit_transform(X)
```

```python
wcss=[]
for k in range(1,10):
    kmc = KMeans(n_clusters=k,random_state=42)
    kmc.fit(X)
    wcss.append(kmc.inertia_)

    
```

#elbow method 


```python
import  matplotlib.pyplot as plt 
plt.scatter(range(1,10),wcss,marker='x')
plt.plot(range(1,10),wcss,color='red')
plt.show()
```

```python
print(kmc.labels_)
print(kmc.cluster_centers_)
```

```python
print(silhouette_score(X,kmc.labels_))
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X[:,0],X[:,1],c=kmc.labels_,s=60)
plt.show()
```

```python
wcss = []
for i in range(1,10):
    kmc = KMeans(n_clusters=i)
    kmc.fit(X)
    wcss.append(kmc.inertia_)

plt.figure(figsize=(7,5))
plt.plot(range(1,10),wcss,marker="o")
plt.show()
```
